# Earth Intelligence Index Engine

## Purpose

The Earth Intelligence Index Engine combines multiple environmental risks into a single standardized Earth Intelligence Index.

The index provides a high-level summary of environmental conditions and supports rapid interpretation for planning and decision-making.

# Import Libraries

## Purpose

Import the libraries required for Earth Intelligence analysis.

```python
from pathlib import Path
import json

import numpy as np
import pandas as pd
```

# Load Risk Assessment Product

## Purpose

Load the Risk Assessment Product generated by the Risk Assessment Engine.

The product contains the standardized thematic risk assessments used to calculate the Earth Intelligence Index.

```python
DATA_DIR = Path(
    "/Users/ShreyaJariwalaMain/_GeoAI_Notebook/Earth-Intelligence-System/data/outputs"
)

risk_assessment_product_file = (

    DATA_DIR

    / "risk_assessment_product.json"

)

with open(

    risk_assessment_product_file,

    "r",

    encoding="utf-8"

) as file:

    risk_assessment_product = json.load(file)

print("Risk Assessment Product loaded successfully.")
```

# Earth Intelligence Framework

## Purpose

Define the contribution of each thematic environmental risk to the Earth Intelligence Index.

Version 1 assigns equal importance to all risk categories.

Future versions may adjust these weights based on application-specific requirements such as flood resilience, climate adaptation, or urban planning.

```python
earth_intelligence_framework = {

    "Flood Risk": 0.25,

    "Heat Risk": 0.25,

    "Vegetation Stress": 0.25,

    "Urban Exposure": 0.25

}

earth_intelligence_framework
```

# Earth Intelligence Index

## Purpose

Calculate the Earth Intelligence Index.

The Earth Intelligence Index combines the thematic environmental risks into a single standardized score ranging from 0 to 100.

Higher values indicate greater overall environmental risk.

```python
earth_intelligence_index = round(

    (

        risk_assessment_product["flood_risk"]["Score"]
        * earth_intelligence_framework["Flood Risk"]

        +

        risk_assessment_product["heat_risk"]["Score"]
        * earth_intelligence_framework["Heat Risk"]

        +

        risk_assessment_product["vegetation_stress"]["Score"]
        * earth_intelligence_framework["Vegetation Stress"]

        +

        risk_assessment_product["urban_exposure"]["Score"]
        * earth_intelligence_framework["Urban Exposure"]

    ),

    2

)

earth_intelligence_index
```

# Earth Intelligence Category

## Purpose

Assign an overall Earth Intelligence category based on the calculated index.

The category provides a simple interpretation of the overall environmental condition of the Area of Interest.

```python
if earth_intelligence_index <= 25:

    intelligence_category = "Low"

elif earth_intelligence_index <= 50:

    intelligence_category = "Moderate"

elif earth_intelligence_index <= 75:

    intelligence_category = "High"

else:

    intelligence_category = "Very High"

intelligence_category
```

# Earth Intelligence Product

## Purpose

Create the standardized Earth Intelligence Product.

The Earth Intelligence Product combines the Earth Intelligence Index, intelligence category, and contributing environmental risks into a single structured product.

This product serves as the final analytical output of the Earth Intelligence System.

```python
from datetime import datetime

earth_intelligence_product = {

    "metadata": {

        "engine": "Earth Intelligence Index Engine",

        "version": "1.0",

        "created_at": datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        )

    },

    "earth_intelligence": {

        "index": earth_intelligence_index,

        "category": intelligence_category

    },

    "contributions": {

        "Flood Risk": risk_assessment_product["flood_risk"]["Score"],

        "Heat Risk": risk_assessment_product["heat_risk"]["Score"],

        "Vegetation Stress": risk_assessment_product["vegetation_stress"]["Score"],

        "Urban Exposure": risk_assessment_product["urban_exposure"]["Score"]

    }

}

earth_intelligence_product
```

# Export Earth Intelligence Product

## Purpose

Export the Earth Intelligence Product generated by the Earth Intelligence Index Engine.

The product is saved as a JSON file and serves as the primary input for the Report Generation Engine.

```python
from pathlib import Path
import json

OUTPUT_DIR = Path(
    "/Users/ShreyaJariwalaMain/_GeoAI_Notebook/Earth-Intelligence-System/data/outputs"
)

earth_intelligence_product_file = (

    OUTPUT_DIR

    / "earth_intelligence_product.json"

)

with open(

    earth_intelligence_product_file,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        earth_intelligence_product,

        file,

        indent=4,

        default=str

    )

print(

    f"Earth Intelligence Product : "

    f"{earth_intelligence_product_file.name}"

)
```

## Purpose

Summarize the final Earth Intelligence results for the selected Area of Interest.

This summary confirms that the Earth Intelligence Product has been successfully created and is ready for report generation.

```python
summary = {

    "Earth Intelligence Index": (

        earth_intelligence_product["earth_intelligence"]["index"]

    ),

    "Category": (

        earth_intelligence_product["earth_intelligence"]["category"]

    ),

    "Flood Risk": (

        risk_assessment_product["flood_risk"]["Category"]

    ),

    "Heat Risk": (

        risk_assessment_product["heat_risk"]["Category"]

    ),

    "Vegetation Stress": (

        risk_assessment_product["vegetation_stress"]["Category"]

    ),

    "Urban Exposure": (

        risk_assessment_product["urban_exposure"]["Category"]

    )

}

for key, value in summary.items():

    print(f"{key:<28}: {value}")
```
