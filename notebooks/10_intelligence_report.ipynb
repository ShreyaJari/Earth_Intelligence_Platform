# Report Generation Engine

## Purpose

The Report Generation Engine consolidates the outputs produced by the Earth Intelligence System into a single standardized report.

The report summarizes the Area of Interest, environmental conditions, risk assessment, and the Earth Intelligence Index.

It provides a concise overview suitable for decision-makers, planners, researchers, and technical stakeholders.

# Import Libraries

## Purpose

Import the libraries required to generate the final Earth Intelligence Report.

```python
from pathlib import Path
import json

from datetime import datetime
```

# Load Products

## Purpose

Load the products generated throughout the Earth Intelligence workflow.

These products provide the information required to generate the final report.

```python
DATA_DIR = Path(
    "/Users/ShreyaJariwalaMain/_GeoAI_Notebook/Earth-Intelligence-System/data/outputs"
)

products = {

    "satellite": "satellite_product.json",

    "terrain": "terrain_product.json",

    "weather": "weather_product.json",

    "land_cover": "land_cover_product.json",

    "environmental": "environmental_indicator_product.json",

    "risk": "risk_assessment_product.json",

    "earth_intelligence": "earth_intelligence_product.json"

}

loaded_products = {}

for key, filename in products.items():

    with open(

        DATA_DIR / filename,

        "r",

        encoding="utf-8"

    ) as file:

        loaded_products[key] = json.load(file)

print(

    f"{len(loaded_products)} products loaded successfully."

)
```

# Report Metadata

## Purpose

Create the metadata for the Earth Intelligence Report.

The metadata identifies the report, records its generation time, and specifies the report version.

```python
report_metadata = {

    "title": "Earth Intelligence Report",

    "version": "1.0",

    "generated_at": datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    ),

    "generated_by": "Earth Intelligence System"

}

report_metadata
```

# Area of Interest Summary

## Purpose

Summarize the Area of Interest used throughout the Earth Intelligence workflow.

This section provides the geographic context for the report.

```python
import geopandas as gpd

aoi_boundary = gpd.read_file(
    DATA_DIR / "aoi.geojson"
)

area_summary = {

    "Coordinate System": aoi_boundary.crs.to_string(),

    "Geometry": aoi_boundary.geometry.iloc[0].geom_type,

    "Area (km²)": round(
        aoi_boundary.to_crs(32642).area.iloc[0] / 1e6,
        2
    ),

    "Bounds": list(
        aoi_boundary.total_bounds
    )

}

area_summary
```

# Environmental Summary

## Purpose

Summarize the key environmental characteristics identified by the Environmental Indicator Engine.

This section provides a concise overview of vegetation, terrain, land cover, and weather conditions within the Area of Interest.

```python
environmental_summary = {

    "Mean NDVI": loaded_products[
        "environmental"
    ]["summary"]["mean_ndvi"],

    "Dominant Land Cover": loaded_products[
        "environmental"
    ]["summary"]["dominant_land_cover"],

    "Mean Elevation (m)": loaded_products[
        "environmental"
    ]["summary"]["mean_elevation"],

    "Mean Temperature (°C)": loaded_products[
        "environmental"
    ]["summary"]["mean_temperature"]

}

environmental_summary
```

# Risk Summary

## Purpose

Summarize the thematic environmental risks identified by the Risk Assessment Engine.

This section provides an overview of the major environmental risks affecting the Area of Interest.

```python
risk_summary = {

    "Flood Risk": loaded_products[
        "risk"
    ]["flood_risk"],

    "Heat Risk": loaded_products[
        "risk"
    ]["heat_risk"],

    "Vegetation Stress": loaded_products[
        "risk"
    ]["vegetation_stress"],

    "Urban Exposure": loaded_products[
        "risk"
    ]["urban_exposure"]

}

risk_summary
```

# Earth Intelligence Summary

## Purpose

Summarize the overall Earth Intelligence assessment.

This section presents the final Earth Intelligence Index and its corresponding intelligence category.

```python
earth_intelligence_summary = {

    "Earth Intelligence Index": (

        loaded_products[
            "earth_intelligence"
        ]["earth_intelligence"]["index"]

    ),

    "Category": (

        loaded_products[
            "earth_intelligence"
        ]["earth_intelligence"]["category"]

    )

}

earth_intelligence_summary
```

# Recommendations

## Purpose

Provide high-level recommendations based on the Earth Intelligence assessment.

Version 1 generates simple recommendations using the overall Earth Intelligence category.

```python
category = earth_intelligence_summary["Category"]

if category == "Low":

    recommendations = [

        "Maintain current environmental management practices.",

        "Continue periodic environmental monitoring."

    ]

elif category == "Moderate":

    recommendations = [

        "Increase monitoring of environmental conditions.",

        "Review land-use planning strategies."

    ]

elif category == "High":

    recommendations = [

        "Prioritize mitigation measures for identified risks.",

        "Increase environmental monitoring frequency.",

        "Review emergency preparedness plans."

    ]

else:

    recommendations = [

        "Immediate environmental assessment is recommended.",

        "Implement risk mitigation strategies.",

        "Increase monitoring and preparedness activities.",

        "Support decision-making with additional field investigations."

    ]

recommendations
```

# Earth Intelligence Report

## Purpose

Create the final Earth Intelligence Report.

The report consolidates the outputs of all Earth Intelligence System modules into a single standardized report for interpretation and decision support.

```python
earth_intelligence_report = {

    "metadata": report_metadata,

    "area_summary": area_summary,

    "environmental_summary": environmental_summary,

    "risk_summary": risk_summary,

    "earth_intelligence_summary": earth_intelligence_summary,

    "recommendations": recommendations

}

earth_intelligence_report
```

# Export Earth Intelligence Report

## Purpose

Export the final Earth Intelligence Report.

The report is saved as a JSON file representing the complete output of the Earth Intelligence System.

```python
OUTPUT_DIR = Path(
    "/Users/ShreyaJariwalaMain/_GeoAI_Notebook/Earth-Intelligence-System/data/outputs"
)

earth_intelligence_report_file = (

    OUTPUT_DIR

    / "earth_intelligence_report.json"

)

with open(

    earth_intelligence_report_file,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        earth_intelligence_report,

        file,

        indent=4,

        default=str

    )

print(

    f"Earth Intelligence Report : "

    f"{earth_intelligence_report_file.name}"

)
```

# Report Generation Engine Summary

## Purpose

Summarize the final Earth Intelligence Report.

This summary confirms that the complete Earth Intelligence workflow has successfully generated the final report.

```python
summary = {

    "Earth Intelligence Index": (

        earth_intelligence_summary[
            "Earth Intelligence Index"
        ]

    ),

    "Category": (

        earth_intelligence_summary[
            "Category"
        ]

    ),

    "Flood Risk": (

        risk_summary[
            "Flood Risk"
        ]["Category"]

    ),

    "Heat Risk": (

        risk_summary[
            "Heat Risk"
        ]["Category"]

    ),

    "Vegetation Stress": (

        risk_summary[
            "Vegetation Stress"
        ]["Category"]

    ),

    "Urban Exposure": (

        risk_summary[
            "Urban Exposure"
        ]["Category"]

    )

}

for key, value in summary.items():

    print(f"{key:<30}: {value}")

print()

print("Earth Intelligence Report generated successfully.")
```
